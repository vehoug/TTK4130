#### Problem 4 - Linked Mechanism

```python
import sympy as sm

# (a) Positions of B and C relative to A in 0-frame

L1, L2 = sm.symbols('L1 L2', real=True)
q1, q2 = sm.symbols('q1 q2', real=True)
dq1, dq2 = sm.symbols('dq1 dq2', real=True)

r1_B_A = sm.Matrix([L1, 0, 0])
r2_C_B = sm.Matrix([L2, 0, 0])

# Rz(q1)
R0_1 = sm.Matrix([
[ sm.cos(q1), -sm.sin(q1), 0],
[ sm.sin(q1),  sm.cos(q1), 0],
[ 0,           0, 1]
])

# Ry(q2)
R1_2 = sm.Matrix([
[ sm.cos(q2), 0, sm.sin(q2)],
[ 0,          1, 0         ],
[-sm.sin(q2), 0, sm.cos(q2)]
])

R0_2 = R0_1 * R1_2

r0_B_A = R0_1 * r1_B_A
r0_C_B = R0_2 * r2_C_B
r0_C_A = r0_B_A + r0_C_B

sm.pprint(r0_B_A)
print()
sm.pprint(r0_C_A)
```

```python
# (b) Angular velocities of AB and BC in 0-frame

omega0_AB = sm.Matrix([0, 0, dq1])
omega0_BC = omega0_AB + R0_2 * sm.Matrix([0, dq2, 0])

sm.pprint(omega0_AB)
print()
sm.pprint(omega0_BC)
```

```python
# (c) Linear velocities of B and C in 0-frame

v0_B = omega0_AB.cross(r0_B_A)
v0_C = v0_B + omega0_BC.cross(r0_C_B)

sm.pprint(v0_B)
print()
sm.pprint(sm.simplify(v0_C))
```
